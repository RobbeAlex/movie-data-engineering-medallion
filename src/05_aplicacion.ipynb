# Aplicación (Model Layer)

## Importación de recursos
En esta celda se cargan las librerías base (Pandas, NumPy) y el módulo centralizado utils.py. La centralización de rutas e importaciones garantiza que el pipeline sea reproducible y que no existan discrepancias en las rutas de acceso a los datos entre los distintos colaboradores del proyecto.

```python
import utils
import pandas as pd
import joblib
```

## Entrenamiento y Evaluación del Modelo Predictivo
Tras el procesamiento del pipeline, se entrena un modelo de Random Forest Regressor para predecir la calificación de las películas (vote_average). Se eligió este modelo por su capacidad para manejar las interacciones complejas entre el presupuesto y la popularidad. Finalmente, los artefactos del modelo y sus métricas de precisión (MAE y R²) se exportan físicamente a la carpeta /models.

```python
modelo, metricas = utils.entrenar_modelo_retail()
```

## CREACIÓN DEL ARCHIVO: modelo_entrenado_seleccionado.pkl
Se utiliza la librería joblib para serializar el modelo entrenado. Guardar el modelo en un archivo físico permite su reutilización en entornos de producción sin necesidad de re-entrenar el sistema, garantizando la portabilidad del producto de datos generado.

```python
joblib.dump(modelo, utils.MODEL_PATH)
```

## CREACIÓN DEL ARCHIVO: metricas.csv
Finalmente, se exportan las métricas de evaluación a un archivo CSV en la carpeta /models. Mantener las métricas junto al modelo es una práctica de MLOps que permite validar rápidamente el rendimiento del sistema y comparar diferentes versiones del modelo basándose en evidencia estadística documentada.

```python
pd.DataFrame([metricas]).to_csv(utils.METRICS_PATH, index=False)
```

```python
print(f"--- Resultados del Modelo de Cine (Clasificación) ---")
print(f"Precisión (Accuracy): {metricas['Accuracy']:.4f}")
print(f"Exhaustividad (Recall): {metricas['Recall']:.4f}")
print(f"Registros procesados: {metricas['Rows_Processed']}")
print(f"✅ Modelo y métricas guardados exitosamente.")
```
